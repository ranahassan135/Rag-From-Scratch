### Word Document Processing

```python
from langchain_community.document_loaders import Docx2txtLoader
from langchain_community.document_loaders import UnstructuredWordDocumentLoader
```

```python
### Method 1 Using Docx2txtLoader
print ("Method 1 Using Docx2txtLoader")
try:
    docx_loader = Docx2txtLoader("data/word_files/proposal.docx")
    docx = docx_loader.load()
    print(f"Loaded {len(docx)} document(s)")
    print(f"Content Preview: {docx[0].page_content[:300]}...")
    print(f"Metadata: {docx[0].metadata}")
except  Exception as e:
    print(f"Error: {e}")
```

```python
### Method 2 
print("Method 2 Using Unstructured")
try:
    unstructured_loader = UnstructuredWordDocumentLoader("data/word_files/proposal.docx" , mode="elements")
    unstructured_docx = unstructured_loader.load()
    
    print(f"Loaded {len(unstructured_docx)} document(s)")
    for i , doc in enumerate(unstructured_docx[:3]):
        print(f"\nElement: {i+1}:")
        print(f"Type {doc.metadata.get('category' , 'unknown')}")
        print(f"Content {doc.page_content[:100]}...")
except Exception as e :
    print(f"Error: {e}")
    
```
